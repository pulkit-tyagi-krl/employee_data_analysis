```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Amit', 'Rahul', 'Priya', 'Neha', 'Rohit',
             'Anjali', 'Karan', 'Simran', 'Vikas', 'Pooja'],
    'Age': [25, 28, 30, 26, 35, 29, 32, 27, 31, 24],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'Sales',
                   'HR', 'IT', 'Sales', 'Finance', 'IT'],
    'Salary': [30000, 35000, 45000, 32000, 50000,
               38000, 48000, 33000, 42000, 28000],
    'Experience': [2, 4, 6, 3, 8, 5, 7, 3, 6, 1]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("First 5 Employees:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nStatistical Analysis:")
print(df.describe())
```

```python
average_salary = df['Salary'].mean()

print("Average Employee Salary:", average_salary)
```

```python
department_count = df['Department'].value_counts()

print(department_count)
```

```python
department_salary = df.groupby('Department')['Salary'].mean()

print(department_salary)
```

```python
df['Department'].value_counts().plot(
    kind='bar',
    title='Employees in Each Department'
)

plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.bar(df['Name'], df['Salary'])

plt.title('Employee Salary Analysis')
plt.xlabel('Employee Name')
plt.ylabel('Salary')

plt.xticks(rotation=45)
plt.show()
```

```python
plt.hist(df['Age'], bins=5)

plt.title('Employee Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Employees')

plt.show()
```

```python
highest_paid = df.loc[df['Salary'].idxmax()]

print("Highest Paid Employee:")
print(highest_paid)
```

```python
lowest_paid = df.loc[df['Salary'].idxmin()]

print("Lowest Paid Employee:")
print(lowest_paid)
```

```python
high_salary = df[df['Salary'] > 40000]

print("Employees earning above 40000:")
print(high_salary)
```

```python
average_experience = df['Experience'].mean()

print("Average Experience:", average_experience)
```

```python
department_salary = df.groupby('Department')['Salary'].mean()

highest_department = department_salary.idxmax()

print("Department with Highest Average Salary:",
      highest_department)

print(department_salary)
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Experience'], df['Salary'])

plt.title('Experience vs Salary')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary')

plt.show()
```

```python
print("===== EMPLOYEE DATA ANALYSIS =====")

print("Total Employees:", len(df))

print("Average Age:", df['Age'].mean())

print("Average Salary:", df['Salary'].mean())

print("Average Experience:", df['Experience'].mean())

print("Highest Salary:", df['Salary'].max())

print("Lowest Salary:", df['Salary'].min())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
plt.figure(figsize=(7,7))

df['Department'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Employee Department Distribution')
plt.ylabel('')
plt.show()
```
